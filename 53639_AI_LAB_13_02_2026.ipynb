Task
1. Create a 3x3 matrix with values ranging from 2 to 10, convert array to a float type. Write a Python program to find the maximum and minimum value of flattened array.


```python
import numpy as np

arr = np.arange(2, 11).reshape(3, 3)

arr = arr.astype(float)

flat_arr = arr.reshape(-1)

maximum_value = np.max(flat_arr)
minimum_value = np.min(flat_arr)

print("Matrix:\n", arr)
print("Flattened Array:", flat_arr)
print("Maximum Value:", maximum_value)
print("Minimum Value:", minimum_value)
```
