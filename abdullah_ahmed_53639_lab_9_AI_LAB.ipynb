```python
# Required Libraries
import numpy as np
from sklearn.model_selection import train_test_split as split_data
from sklearn.neighbors import KNeighborsClassifier as KNN
from sklearn.metrics import accuracy_score as acc_score
from sklearn.datasets import load_iris, make_classification

# -------------------------
# FUNCTION TO RUN KNN
# -------------------------
def evaluate_knn(X_train, X_test, y_train, y_test, k_list):

    for value in k_list:

        classifier = KNN(n_neighbors=value)
        classifier.fit(X_train, y_train)

        predictions = classifier.predict(X_test)
        accuracy = acc_score(y_test, predictions)

        print(f"Neighbors: {value} --> Accuracy: {round(accuracy, 4)}")


# -------------------------
# DATASET 1 : IRIS
# -------------------------
iris_data = load_iris()

features_1 = iris_data.data
labels_1 = iris_data.target

X_train1, X_test1, y_train1, y_test1 = split_data(
    features_1, labels_1, test_size=0.30, random_state=0
)

print("----- IRIS DATA RESULTS -----")

k_values = [1, 3, 5, 7]
evaluate_knn(X_train1, X_test1, y_train1, y_test1, k_values)


# -------------------------
# DATASET 2 : SYNTHETIC DATA
# -------------------------
X_syn, y_syn = make_classification(
    n_samples=300,
    n_features=5,
    n_classes=3,
    n_clusters_per_class=1,
    random_state=0
)

X_train2, X_test2, y_train2, y_test2 = split_data(
    X_syn, y_syn, test_size=0.30, random_state=0
)

print("\n----- GENERATED DATA RESULTS -----")

evaluate_knn(X_train2, X_test2, y_train2, y_test2, k_values)
```
